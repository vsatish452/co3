```python
%sql

-- 1. Source table
CREATE TABLE IF NOT EXISTS workspace.bronze.emp_source (
    emp_id INT,
    emp_name STRING,
    department STRING,
    salary DECIMAL(10,2),
    city STRING,
    last_updated_date TIMESTAMP,
    operation STRING
)
USING DELTA;
```

```python
%sql

-- 2. Bronze table
CREATE TABLE IF NOT EXISTS workspace.bronze.emp_bronze (
    emp_id INT,
    emp_name STRING,
    department STRING,
    salary DECIMAL(10,2),
    city STRING,
    last_updated_date TIMESTAMP,
    operation STRING
)
USING DELTA;
```

```python
%sql

-- 3. Silver target table
CREATE TABLE IF NOT EXISTS workspace.silver.emp_silver (
    emp_id INT,
    emp_name STRING,
    department STRING,
    salary DECIMAL(10,2),
    city STRING,
    last_updated_date TIMESTAMP
)
USING DELTA;
```

```python
%sql

-- 4. Watermark table
CREATE TABLE IF NOT EXISTS workspace.silver.etl_watermark (
    table_name STRING,
    last_load_date TIMESTAMP
)
USING DELTA;
```

```python
%sql

-- 5. SCD Type 2 table
CREATE TABLE IF NOT EXISTS workspace.silver.emp_scd2 (
    emp_id INT,
    emp_name STRING,
    department STRING,
    salary DECIMAL(10,2),
    city STRING,
    effective_start_date TIMESTAMP,
    effective_end_date TIMESTAMP,
    is_current BOOLEAN
)
USING DELTA;
```
