```python
from pyspark.sql.functions import col, lit, current_timestamp, when, concat
```

```python
Delivery_df = spark.read.table("workspace.bronze.co2Delivery")
Customer_df = spark.read.table("workspace.bronze.co2Customer")
Material_df = spark.read.table("workspace.bronze.co2Material")
Distance_df = spark.read.table("workspace.bronze.co2Distance")
Emission_df = spark.read.table("workspace.bronze.co2emission_factor")
display(Delivery_df)
display(Customer_df)
display(Material_df)
display(Distance_df)
display(Emission_df)

```

```python
# Joining Delivery with Material table using left join
delivery_material_df = Delivery_df.alias("dlv")\
                      .join(Material_df.alias("mlt"),col("dlv.MTL_NO")==col("mlt.MTL_NO"),"left")\
                      .select(
                              Delivery_df["*"],
                              col("mlt.MTL_WT").alias("MTL_WT"),
                              col("mlt.MTL_WT_UNIT").alias("MTL_WT_UNIT")
                              )
                        
#display(delivery_material_df)

# Joining delivery_material_df with Customer table using left join
delivery_m_customer_df = delivery_material_df.alias("dlv_mlt")\
                        .join(Customer_df.alias("cust"),col("dlv_mlt.CUST_NO")==col("cust.CUST_NO"),"left")\
                        .select(
                            delivery_material_df["*"], 
                            col("cust.SHIP_TO_CUST_NM").alias("SHIP_TO_CUST_NM"),
                            concat(

                            col("CUST_CITY"),lit(","), col("CUST_CNTRY"),lit(","),col("SHIP_TO_POSTL_CD")).alias ("CUST_address")
                            )
                        
display(delivery_m_customer_df)
```

```python
# Joining delivery_m_customer_df with Distance table using left join 
delivery_mc_distance_df = delivery_m_customer_df.alias("dcm")\
                         .join(Distance_df.alias("dist"),
                               (col("dcm.SHIP_FROM_POSTL_CD") == col("dist.SHIP_FROM_POSTL_CD")) &
                               (col("dist.SHIP_TO_POSTL_CD") == col("dist.SHIP_TO_POSTL_CD")), "left")\
                         .select(
                                delivery_m_customer_df["*"],
                                col("dist.DISTANCE_KM").alias("DISTANCE_KM")
                              ) 
                            
display(delivery_mc_distance_df) 
```

```python
# Joining delivery_mc_distance_df with Emission table using left join
delivery_mc_emission_df = delivery_mc_distance_df.alias("dd")\
                          .join(Emission_df.alias("emi"),
                               (col("dd.COUNTRY") == col("emi.COUNTRY")) &
                               (col("dd.TRANSPORT_MODE") == col("emi.TRANSPORT_MODE")) &
                               (col("dd.VEHICLE_TYPE") == col("emi.VEHICLE_TYPE")) &
                               (col("dd.VEHICLE_CAPACITY") == col("emi.VEHICLE_CAPACITY")) &
                               (col("dd.FUEL_TYPE") == col("emi.FUEL_TYPE")),"left")\
                          .select(
                                  delivery_mc_distance_df["*"],
                                  col("emi.EMISN_FACTR").alias("EMISN_FACTR")
                                )
                            
display(delivery_mc_emission_df)
```

```python
#Calculating weight in tons (casting MTL_WT and DLVRD_QTY to double since they are stored as strings)
Weight_in_tons_df= delivery_mc_emission_df.withColumn("weight_in_ton", 
                               when(col("MTL_WT_UNIT") == "LBS",col("MTL_WT").cast("double") * col("DLVRD_QTY").cast("double") * 0.453592 * 0.001)
                               .when(col("MTL_WT_UNIT") == "G", col("MTL_WT").cast("double") * col("DLVRD_QTY").cast("double") * 0.001 * 0.001)
                               .when(col("MTL_WT_UNIT") == "OZ", col("MTL_WT").cast("double") * col("DLVRD_QTY").cast("double") * 0.0283495 * 0.001)
                               .when(col("MTL_WT_UNIT") == "KG", col("MTL_WT").cast("double") * col("DLVRD_QTY").cast("double") * 0.001)
                               .otherwise(None))
display(Weight_in_tons_df)    
```

```python
# Calculating CO2 emission in tons 
CO2_emission_tons = Weight_in_tons_df.withColumn("co2_emission_in_ton", 
                    (col("weight_in_ton") * col("DISTANCE_KM") * col("EMISN_FACTR"))/1000)
display(CO2_emission_tons)                               
```

```python
# Selecting all columns as final dataframe
CO2_Emission_df = CO2_emission_tons.select(col("DLVRY_NO").alias("Delivery_no"),
                                           col("SHIP_FROM_CNTRY").alias("Country"),
                                           col("GI_DATE").alias("Goods_issue_date"),
                                           col("CUST_NO").alias("Customer_no"),
                                           col("SHIP_TO_CUST_NM").alias("Customer_name"),
                                           col("CUST_address").alias("Customer_address"),
                                           col("MTL_NO").alias("Material_no"),
                                           col("MTL_WT").alias("Material_weight"),
                                           col("MTL_WT_UNIT").alias("Material_weight_unit"),
                                           col("DLVRD_QTY").alias("Delivery_quantity"),
                                           col("weight_in_ton").alias("Weight_in_ton"),
                                           col("DISTANCE_KM").alias("Distance_in_km"),
                                           col("TRANSPORT_MODE").alias("Transport_mode"),
                                           col("VEHICLE_TYPE").alias("Vehicle_type"),
                                           col("VEHICLE_CAPACITY").alias("Vehicle_capacity"),
                                           col("FUEL_TYPE").alias("Fuel_type"),
                                           col("EMISN_FACTR").alias("Emission_factor"),
                                           col("co2_emission_in_ton").alias("CO2_emission_in_ton"),
                                           col("L1_upd").alias("Update_timestamp"))\
                                           .orderBy(col("Delivery_no").asc())
display(CO2_Emission_df)                                           
                                           
```

```python
CO2_Emission_df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.CO2_Emission")
```
